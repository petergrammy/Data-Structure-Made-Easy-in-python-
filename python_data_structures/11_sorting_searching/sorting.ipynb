## Select

In bubble we compare the number and get 1 biggest each time.

In select, we select a biggest and put it at the 'relative end'.

*the following code will be only algorithm itself, find the complete class at the end of the chapter.*



```python
def select(self):
    array1=self.array
    for passnum in range(len(array1)-1,0,-1):
        max_position = 0
        for i in range(passnum):
            if array1[i]>array1[max_position]:
                max_position = i   
        array1[passnum],array1[i]=array1[i],array1[passnum]
```

## Insert

In insert we start from one element, and insert elements one by one.

One element is compared to each element in the sorted array, the index of the posistion matters most.

```python
def insert(self):
    array1 = self.array
    for index in range(1,len(array1)):
        position = index
        value = array1[index]
        while position>0 and array1[index-1]<value:
            array1[position]=array1[position-1]
            position -= 1
        array1[position]=value
```

When you actually write this algorithm, you'll find it harder than expected. But, just as any other algorithm, you're just not familiar with its unique logic. Instead of operate the array, we here control the position index, so that we avoid moving other elements when inserting one.

But how do you actually achieve it?

In any algorithm with an O($n^2$), it basicly means that two 'for' or 'while' are included. One puts the operation into action, the other makes sure every element go through the first one. In this algorithm, the 'operation while' goes like:
- pretend the chosen element to be the biggest.
- compare it with the 'n-1'.
- if that's the case, the element shall not be moved, the while stops, and the position is given the current value.
- if it's smaller than its prior element, we **move the prior element to the current position**, and perform the second step.

You'll find we extract the value, compare it, after which insert it to the position.

```python
def partition(list,first,last):
    pivot_value = list[first]
    leftmark = first+1
    rightmark = last
    done = False

    while not done:
        while list[leftmark]<=pivot_value and leftmark<=rightmark:
            leftmark+=1
        while list[rightmark]>=pivot_value and leftmark<=rightmark:
            rightmark-=1

        if leftmark>rightmark:
            done = True
        else:
            list[leftmark],list[rightmark]=list[rightmark],list[leftmark]

    list[first],list[rightmark]=list[rightmark],list[first]
    
    return rightmark

def quickhelper(list,first,last):
    if first < last:
        split_point = partition(list,first,last)
        quickhelper(list,first,split_point-1)
        quickhelper(list,split_point+1,last)

def quicksort(self):
    quickhelper(self.array,0,len(self.array)-1)
    
```
