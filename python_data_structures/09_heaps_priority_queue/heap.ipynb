## Practice: To Build a MinHeap

```python
class MinHeap:
    def __init__(self):
        self.heap=[0]
        self.heapSize=0
```

As you can see, the __init()__ function looks pretty strange, because it pushes the element 0 directly into the heap, which is not common in any initial function.

Doing so is to the mathematical behind heap algorithm.

As we are tring to build a MinHeap by actually building an array,the Minheap in array form goes with a rule like this:

- if the index for the parent is **i**;
- then the index for the left child is **i*2**;
- and the right child is **i*2+1**.

When the index shows up, we want it to start from 1, not 0. Hence we push 0 into the heap right when a heap is built.right?

```python
class MinHeap:
    def __init__(self):
        self.heap=[0]
        self.heapSize=0

    def percUp(self,i):
        while i//2 > 0:
            if self.heap(i)<self.heap(i//2):
                self.heap(i),self.heap(i//2)=self.heap(i//2),self.heap(i)
            i = i//2
```

**Note** that, *i//2* means i divided by 2 to get an integer, and we don't care if there's anything left.

so if 5//2, it's 2. 8//2, it's 4.

By percUp(), we're capable of 'checking' if any given element in the heap legal or not.
