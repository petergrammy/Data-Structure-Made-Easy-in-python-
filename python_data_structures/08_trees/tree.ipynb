````python
def BinaryTree(r):
    return [r,[],[]]

def insertLeftSubTree(root,subtree):
    t = root.pop(1)
    if len(t)>1:
        root.insert(1,[subtree,t,[]])
    else:
        root.insert(1,[subtree,[],[]])
    return root

def insertRightSubTree(root,subtree):
    t = root.pop(2)
    if len(t)>1:
        root.insert(2,[subtree,[],t])
    else:
        root.insert(2,[subtree,[],[]])

    return root
````

## insert a subtree

A tree in list form goes like this:

```python
[r,[],[]]
```
where 'r' represents root, which itself is a node,
and one [] for left subtree, another [] for right subtree.

When you write `insert()`, point out the position index, and then write what you want to insert. So if you insert a left subtree, the position index would be 1, the right would be 2.

Take `insertLeftSubtree(root,subtree)` as example:

Here we need to consider two different situations:
- one.The left subtree is originally complete, the tree itself may like `[r,[left_root1,[left_root1],[left_root2]],[]]`.
In this case, when you perform 
```python
t = root.pop(1)
if len(t)>1:
```
the `len(t)` is the length of origin list reps left subtree, in this case, `[left_root1,[left_root1],[left_root2]]`, the length of which equals to 3.

To this, we perform 
```python
root.insert(1,[subtree,t,[]])
```

doing so means we put the origin t, the `[left_root,blah,blah...]`, *packed* into the left subtree of `subtree`, which, the`insertLeftSubtree(root,subtree)` stands, as well as the `root.insert(1,[subtree,t,[]])` stands.

As a matter of fact, long as `len(t)>1`, we cannot guarantee that the space inside the origin tree is enough.

- two.The left subtree is full of space, like `[]`, and the tree may be `[r,[],[right_root,[],[]]]`

so, after we pop the `root(1)`, the length is 0. So we can directly perform the insert:
```python
root.insert(1,[subtree,[],[]])
```


## Set Value

````python
def set_value_for_a_node(root,node_type,value):
    if node_type == 'root':
        root[0] = value
    elif node_type == 'left':
        root.insertLeftSubTree(root,[value.append(root[1].pop(0))])
    else:
        root.insertRightSubTree(root,[value.append(root[2].pop(0))])

````
