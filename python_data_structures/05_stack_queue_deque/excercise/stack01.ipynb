
给出一个由(和)字符组成的字符串。求出它最长的是合法括号序列的子串的长度，同时求出最长子串的个数。

如果一个括号序列插入若干+和1后，可以得到一个正确的数学表达式，那么它被称为“合法”的。

例如，序列(())()，()和(()(()))是合法的，但)(，(()和(()))(不是。

输入

一行包含一个非空字符串，由“ (”和“ )” 字符组成，长度为 n 。

输出

输出最长合法括号序列的子串的长度，和最长子串的个数。
如果不存在这种子串，输出唯一的一行，包含 0 1

样例

input

)((())))(()())

output

6 2


```python
## 时间复杂度分析
# TODO: 你的时间复杂度分析从这里开始


from typing import Tuple
def longest_valid_parentheses(s: str) -> Tuple[int, int]:
    # TODO: 你的代码
    stack=[-1]
    max_length=0
    count=0
    for i,c in enumerate(s):
        if c=='(':
            stack.append(i)
        else:
            stack.pop()
            if not stack:
                stack.append(i)
            else:
                length=i-stack[-1]
                if length>max_length:
                    max_length=length
                    count=1
                elif length==max_length:
                    count+=1

    # 请自行修改返回值为合适的变量
    if max_length==0:
        count=1
    return max_length,count


# 从标准输入读取字符串
s = input().strip()

# 调用函数解决问题
max_length, count = longest_valid_parentheses(s)
# 输出结果
print(f"{max_length} {count}")
```
