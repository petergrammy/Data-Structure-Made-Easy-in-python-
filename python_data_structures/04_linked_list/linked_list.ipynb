```python
## 时间复杂度分析
# TODO: 你的时间复杂度分析从这里开始
# 对长度为m和n的两个链表进行了一次遍历，因此时间复杂度应该是O(m+n)
from typing import List

class ListNode:
    def __init__(self, value=0, next=None):
        self.value = value
        self.next = next

def mergeLists(nums1: ListNode, m: int, nums2: ListNode, n: int) -> ListNode:
    #TODO: 你的代码
    a=nums1.next
    b=nums2.next
    lst=[]
    while a is not None or b is not None:
        if b is None or (a is not None and a.value <= b.value):
            lst.append(a.value)
            a=a.next
        else:
            lst.append(b.value)
            b=b.next

    # 输出应该是一个ListNode,请自行修改返回值为合适的变量
    return list_from_values(lst) #请将[1, 2, 3, 4, 5]改为合并后的list

# 代码自测
def list_from_values(values):
    head = ListNode()
    curr = head
    for value in values:
        curr.next = ListNode(value)
        curr = curr.next
    return head

def print_list(node):
    now = node.next
    while now:
        print(now.value, end=' ')
        now = now.next
    print()


# 读取输入
n, m = map(int, input().split())
values1 = list(map(int, input().split()))
values2 = list(map(int, input().split()))

# 创建链表
list1 = list_from_values(values1)
list2 = list_from_values(values2)

# 合并链表
merged = mergeLists(list1, n, list2, m)

# 打印结果
print_list(merged)
```
