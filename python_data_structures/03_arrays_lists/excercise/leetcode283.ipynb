# Leetcode283: Move 0's to the end 

Input: an array
Output: an array, where the 0's moved to the end of it

```python
class Solution:
    def moveZeroes(self,nums:List[int])->None:
        fast=0
        for slow in range(len(nums)):
            if nums[slow]!=0:
                nums[fast],nums[slow]=nums[slow],nums[fast]
                fast+=1
        
```

```python
nums=[0,0,2]
lst=[]
for i in range(len(nums)):
    if nums[i]==0:
        lst.append(i)
for i in lst:
    del nums
```

```python
nums=[0,0,5,6,7]
count=0
while 0 in nums:
    nums.remove(0)
    count+=1
for i in range(count):
    nums.append(0)
print(nums,count)
```
