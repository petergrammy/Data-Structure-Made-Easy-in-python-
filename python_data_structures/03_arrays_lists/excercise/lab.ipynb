## 题2
后缀表达式指：式中不使用括号，运算符号无优先级。运算符都放在两个运算对象之后，所有计算按运算符号出现的顺序，严格地由左而右进行。

如：3*(5-2)+7 对应的后缀表达式为：3 5 2 - * 7 + @。在该式中，@ 为表达式的结束符号。

输入

一行字符串，可能包括数字、加减乘除、结束符@。

输出

一个数，为表达式的结果。

```python
def evaluate_postfix(expression):
    #your code here
    stack = []
    comma = ['-', '*', '+', '/', '@']


    for i in expression:
        if i==' ':
            continue
        if i == '@':
            break

        if i not in comma:
            stack.append(int(i))

        else:
            a = stack.pop()
            b = stack.pop()

            if i == '+':
                stack.append(b + a)
            elif i == '-':
                stack.append(b - a)
            elif i == '*':
                stack.append(b * a)
            elif i == '/':
                stack.append(b / a)

    return stack[-1]


# 读取输入

expression = input().strip()

# 输出结果
print(evaluate_postfix(expression))
```

## 题3
有两个单调递增的线性表，希望你以顺序表的形式存储，然后将两个线性表合并成一个有序的线性表。

输入

第一行两个整数 n ， m ，表示第一个线性表的大小和第二个线性表的大小

第二行 n 个整数，表示第一个线性表的 n 个元素

第三行 m 个整数，表示第二个线性表的 m 个元素

输出

一行，n+m 个整数，表示合并后的线性表

```python
n, m = map(int, input().split())
list1 = list(map(int, input().split()))
list2 = list(map(int, input().split()))

node1 = 0
node2 = 0
lst = []

while node1 < n or node2 < m:
    if node2 == m:
        lst.append(list1[node1])
        node1 += 1
    else:
        if node1 == n:
            lst.append(list2[node2])
            node2 += 1
        elif list1[node1] <= list2[node2]:
            lst.append(list1[node1])
            node1 += 1
        else:
            lst.append(list2[node2])
            node2 += 1

for _ in lst:
    print(_)
```

## 题4
有一个乱序的线性表A，希望你完成这样的任务：以A的第一个元素为链表B的初始元素，遍历A的元素，将各个元素插入到B中合适的位置，以使整个链表单调递增。并输出链表B最终的值。

题意即在链表上实现插入排序。具体用单链表还是双链表实现不做要求。

输入

第一行一个整数 n ，表示线性表A的元素个数。

第二行 n 个整数，表示线性表A的 n 个元素。

输出

一行，n 个整数，表示链表B的内容

```python
# your code here
class ListNode:
    def __init__(self, val):
        self.val = val
        self.next = None


n = int(input())
values = list(map(int, input().split()))

head = ListNode(values[0])

for x in values[1:]:
    new_node = ListNode(x)

    
    if x < head.val:
        new_node.next = head
        head = new_node

    else:
        curr = head

        while curr.next is not None and curr.next.val <= x:
            curr = curr.next

        
        new_node.next = curr.next
        curr.next = new_node

curr = head
while curr is not None:
    print(curr.val, end=" ")
    curr = curr.next
```

## 栈2
给定A序列和B序列，它们的长度都为 n ，且序列内元素的取值互不相同，都在 1 到 n 之间，即 1 到 n 的每个数在两个序列中都分别出现了一次。询问当A序列为入栈序列时，B序列是否可能为出栈序列

输入

第一行一个整数 n 。

第二行 n 个整数，表示A序列。

第三行 n 个整数，表示B序列。

输出

一行，0或1，0表示不可能，1表示可能

```python
def is_possible_output(a, b):
    # your code here
    stack=[]
    j=0
    for i in range(n):
        stack.append(a[i])
        while stack and i<n and stack[-1]==b[j]:
            stack.pop()
            j+=1

    return j==n
# 读取输入
n = int(input())
a = list(map(int, input().split()))
b = list(map(int, input().split()))

# 输出结果
print(1 if is_possible_output(a, b) else 0)
```
