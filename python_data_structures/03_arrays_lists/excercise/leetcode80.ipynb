# Leetcode 80: Remove Duplicates from Sorted Arrays II
题目描述

给你一个非严格递增排列的数组 nums，请你原地删除重复出现的元素，使每个元素最多出现两次，返回删除后数组的新长度。

不要使用额外的数组空间，你必须在原地修改输入数组并在使用 O(1) 额外空间的条件下完成。

说明

nums 是已经排好序的（非递减）。你只需要保证每个元素出现不超过两次。

不要求考虑数组中超出新长度后面的元素。

示例 1

输入: nums = [1,1,1,2,2,3]
输出: 5, nums 前五个元素变为 [1,1,2,2,3]
解释: 函数应返回新长度 5，且原数组的前五个元素要被修改为 1, 1, 2, 2, 3。

示例 2

输入: nums = [0,0,1,1,1,1,2,3,3]
输出: 7, nums 前七个元素变为 [0,0,1,1,2,3,3]

提示

- $1 <= nums.length <= 3 * 10^4$
- $-10^4 <= nums[i] <= 10^4$
- nums 已按非递减顺序排序

---

思考题

1. 你能只用 两个指针 完成吗？一个快指针负责"遍历看每个元素"，一个慢指针负责"往哪放"。
2. 关键是:遍历时怎么判断一个元素该不该保留？提示：如果 nums[fast] 和 nums[slow-2] 相等，说明什么？
3. 你的慢指针"等待写入"的位置和已保留的个数之间是什么关系？

---

```python
class Solution:
    def removeDuplicates(self, nums: list[int]) -> int:
        fast=0
        slow=0
        count=0
        if not nums:
            return 0
        if len(nums)<3:
            return len(nums)
        nums.append(None)
        while nums[fast]!=None:
            while nums[fast]==nums[slow]:
                fast+=1
                count+=1
            if count>2:
                nums[slow+2:]=nums[fast:]+(count-2)*[None]
                count=2
            if count!=0:
                fast=slow+count
            if count==0:
                fast+=1
            slow=fast
            count=0
        return slow
```

```python
class Solution:
    def removeDuplicates(self, nums: list[int]) -> int:
        if not nums:
            return 0
        fast = 0
        slow = 0
        count = 0
        while fast < len(nums):
            while fast < len(nums) and nums[fast] == nums[slow]:
                fast += 1
                count += 1
            if count > 2:
                nums[slow+2:] = nums[fast:]   # (count-2)*[] 恒等于 []，去掉
            fast = max(fast, slow + count)    # 修正 fast 的位置
            slow += count                     # slow 移到下一个连续段
            count = 0
        return slow
```
